# Lecture - DevOps and CI/CD Review

## Overview

In this lecture, you will review the DevOps and CI/CD foundations that the rest of this course builds on. These topics are prerequisites, so the review moves quickly and establishes a shared vocabulary before you begin deploying Databricks projects with Declarative Automation Bundles (DABs).

## Learning Objectives

By the end of this lecture, you will be able to:
1. **Describe** the core concepts and benefits of DevOps and explain how they extend to data engineering as DataOps.
2. **Explain** Continuous Integration and Continuous Delivery/Deployment (CI/CD), and distinguish continuous delivery from continuous deployment.
3. **Describe** the role of automated testing across unit, integration, and system tests.
4. **Explain** how environments and data are isolated across DEV, STAGE, and PROD.
5. **Identify** the tools available to deploy Databricks projects: the REST API, the Databricks SDKs, and the Databricks CLI.

#### What is DevOps?
DevOps is the union of two worlds that traditionally worked separately.

- **Software engineering best practices** contribute version control, automated testing, and repeatable builds.
- **IT operations** contributes deployment, monitoring, and infrastructure management.

The overlap in the middle is DevOps. A culture and a set of practices that get those two groups working together instead of throwing work over a wall.

#### Why it matters
- **Faster deployments.** Automation removes slow, manual handoffs, so changes reach users sooner.
- **Improved collaboration.** Dev and ops share ownership of the whole lifecycle rather than blaming each other when something breaks.
- **Enhanced reliability.** Automated testing and consistent deployments mean fewer surprises in production.
- **Better scalability.** Repeatable, automated processes scale far better than manual ones.

In short, DevOps is a way to build and deliver software quickly and reliably by bridging the gap between development and operations..

</details>

##### EXPAND FOR ADDITIONAL NOTES

<details>

- CI involves regularly merging code changes from multiple contributors into a central repository and running automated tests to ensure code quality. Tests that fail do not make it into your source code.
- For example, imagine three developers working on a project, implementing updates or fixes.
- The developers push their code to a version control system, which automatically runs tests before the commit is made. This ensures that any issues with the code are caught. If the tests fail, the commit is stopped. If the tests pass, the changes are implemented.
- The way testing and commits to version control are handled depends on the branching strategy defined by your organization. Deciding on the right branching and testing strategy is crucial for maintaining quality and smooth workflows.
- Here are four key benefits of Continuous Integration:
- First, Early Detection of Issues—by integrating code often, bugs and conflicts are caught early, making them easier to fix.
- Second, Faster Development Cycle—frequent integration speeds up the delivery of new features and fixes.
- Third, Improved Collaboration and Code Quality—regular integration leads to cleaner, more modular code and better teamwork.
- Finally, Automated Testing and Validation—automated tests run with each integration to ensure the code is stable and works with existing features.
- These benefits—early issue detection, faster delivery, better collaboration, and automated testing—make Continuous Integration essential for smooth development.

</details>

##### EXPAND FOR ADDITIONAL NOTES

<details>

- The testing pyramid categorizes tests into unit, integration, and system tests. Testing your code is essential.
- Unit tests form the base: they test individual functions or methods in isolation, run quickly and automatically, and provide the broadest coverage at low cost &mdash; for example, testing a custom pyspark function.
- Integration tests sit in the middle: they test the interaction between components or systems, and are slower and costlier than unit tests &mdash; for example, notebooks, declarative pipelines, or jobs.
- System tests sit at the top: they test the entire application end to end in a production-like environment, and are the slowest and most expensive &mdash; for example, an end-to-end data pipeline in a Workflow.

</details>
