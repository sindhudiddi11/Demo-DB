# 16 Bonus - Using VS Code with Databricks

## Overview

Up to this point you've built and deployed bundles entirely from inside the Databricks workspace. In real production work, most teams author bundles in **VS Code** with the **Databricks extension**, which gives you syntax highlighting, schema validation, autocomplete for **databricks.yml**, and one-click bundle deploy/run from the editor.

This bonus walkthrough shows you how to authenticate VS Code against your Databricks workspace using the workspace URL and a Personal Access Token (PAT), so you can switch from notebook-based bundle work to a local editor.

## Learning Objectives

By the end of this demonstration, you will be able to:

1. **Find the Databricks workspace URL** the VS Code extension needs for authentication.
2. **Generate a Personal Access Token (PAT)** from the workspace **User Settings**.
3. **Open the VS Code lab environment** and authenticate it with the workspace URL and PAT.

## A. Open a Text Editor

1. Open a simple text editor on your computer.

2. Follow the steps below to paste the following information into the file so you can use it later when authenticating with VS Code:

   - Your Databricks workspace URL

   - Your Personal Access Token (PAT)

## B. Get Your Databricks Workspace URL

1. Run the cell below to print your workspace URL. Leave the output visible. You'll paste it into your text editor and again into VS Code.

```python
lab_databricks_url = f'{spark.conf.get("spark.databricks.workspaceUrl")}/'
print(lab_databricks_url)
```

## C. Generate a Personal Access Token (PAT)

Generate a PAT in the workspace so VS Code can authenticate.

1. Click on your username in the top bar, right-click on **User Settings** in the drop-down, and select **Open in a New Tab**.

2. In **Settings**, select **Developer**, then to the left of **Access tokens**, select **Manage**.

3. Click **Generate new token**.
   
4. Set **Scope** to **Other APIs**.

5. Set **API Scope** and check **all APIs (not recommended)**.
   - For training purposes. Follow your organization's security and governance guidelines when defining token permissions.

6. Click **Generate**.

7. Copy the displayed token to the clipboard. **You will not be able to view the token again.** If you lose it, you'll need to delete it and create a new one.

8. Paste the PAT into your text editor below the workspace URL.
