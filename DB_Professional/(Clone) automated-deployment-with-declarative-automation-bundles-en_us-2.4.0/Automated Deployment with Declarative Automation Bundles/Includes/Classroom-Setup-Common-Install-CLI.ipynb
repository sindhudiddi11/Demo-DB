```python
%sh
############################################################
## Only works on a Dedicated (formerly: Single user) cluster with policy Unrestricted
## Has not been tested elsewhere.
############################################################

# Remove any existing Databricks CLI binary from the root user's local bin directory
# -f forces removal and avoids errors if the file does not exist
sudo rm -f /root/bin/databricks

# Remove any existing Databricks CLI binary from the system-wide bin directory
# This ensures no old or conflicting versions remain
sudo rm -f /usr/local/bin/databricks

# Download the official Databricks CLI install script from GitHub
# - `curl -fsSL https://raw.githubusercontent.com/databricks/setup-cli/v0.240.0/install.sh | sh`: 
#   - Downloads the `install.sh` script from a specific GitHub URL using `curl`.
#   - The options used:
#     - `-f`: Fail silently on server errors.
#     - `-s`: Silent mode (no progress or error messages).
#     - `-S`: Show errors if they occur.
#     - `-L`: Follow redirects if any.
#   - Pipes (`|`) the downloaded script directly to the shell (`sh`), which executes it.
# Pipe the script directly into the shell to execute the installation
curl -fsSL https://raw.githubusercontent.com/databricks/setup-cli/v0.298.0/install.sh | sh

# Move the installed CLI binary from the root bin directory to a system-wide location
# This makes the `databricks` command available globally on the cluster
# Moves the `databricks` executable from `/root/bin/` to `/usr/local/bin/`, making it available to users without having to specify the location during each cell execution in the notebook.
sudo mv /root/bin/databricks /usr/local/bin/databricks

# The cell will return the note: *Installed Databricks CLI v0.2XX.0 at /root/bin/databricks.*
```
